```python
import pandas as pd

# Load the original dataset
df = pd.read_csv("project1_algorithmic_pricing_synthetic_dataset.csv")

# Load the prepared datasets
daily_wide = pd.read_csv("daily_wide_revenue.csv")
weekly_wide = pd.read_csv("weekly_wide_revenue.csv")

print("DOCUMENTATION CHECK")
print("-" * 35)

print("Original dataset:", df.shape)
print("Daily revenue table:", daily_wide.shape)
print("Weekly revenue table:", weekly_wide.shape)

print("\nOriginal dataset columns:")
print(df.columns.tolist())

print("\nDaily table columns:")
print(daily_wide.columns.tolist())

print("\nWeekly table columns:")
print(weekly_wide.columns.tolist())
```

```python
data_dictionary = """
# Data Dictionary

## Project
Causal Inference & Synthetic Control for Algorithmic Pricing

## Dataset Source
The dataset is synthetic/simulated teaching data, not real company data.

## Original Dataset
- File: project1_algorithmic_pricing_synthetic_dataset.csv
- Rows: 540
- Columns: 11
- Date range: 2025-01-01 to 2025-06-29
- Regions: North, South, West
- Treatment region: North
- Intervention date: 2025-05-01

## Original Dataset Columns

| Column | Description |
|---|---|
| date | Date of the observation |
| region | Region represented by the observation |
| product_category | Product category |
| our_price | Price set by the business |
| competitor_price | Competitor's price |
| units_sold | Units sold |
| revenue | Recorded revenue for the observation |
| marketing_spend | Marketing expenditure |
| promotion_active | Indicates whether a promotion is active |
| treatment | Indicates treatment status for the observation |
| period | Pre-treatment or post-treatment period |

## Prepared Dataset 1: Daily Revenue
- File: daily_wide_revenue.csv
- Rows: 180
- Columns: 4
- Columns: date, North, South, West
- Each row represents one date.
- North, South and West contain the summed recorded revenue for each region on that date.

## Prepared Dataset 2: Weekly Revenue
- File: weekly_wide_revenue.csv
- Rows: 26
- Columns: 4
- Columns: week_start, North, South, West
- Each row represents one week.
- Weeks start on Monday.
- North, South and West contain the summed recorded revenue for each region during that week.
- The first week starts on 2024-12-30 because weekly periods start on Monday.

## Intended Use
The prepared revenue tables support regional revenue comparison and further causal analysis using Synthetic Control and related methods.

## Validation Summary
- Original dataset shape: 540 rows x 11 columns
- Daily revenue table shape: 180 rows x 4 columns
- Weekly revenue table shape: 26 rows x 4 columns
- Missing values in prepared tables: 0
- Duplicate dates/weeks in prepared tables: 0
"""

with open("data_dictionary.md", "w", encoding="utf-8") as file:
    file.write(data_dictionary.strip() + "\n")

print("data_dictionary.md created successfully!")
```

```python
readme_content = """
# Causal Inference & Synthetic Control for Algorithmic Pricing

## Project Overview
This project studies the impact of a simulated algorithmic pricing intervention
using causal inference methods.

The objective is to compare revenue in the treated region with comparison regions
and prepare data for further causal analysis.

## Dataset Information
- Dataset: project1_algorithmic_pricing_synthetic_dataset.csv
- Source: Synthetic/simulated teaching dataset
- Observations: 540 rows
- Columns: 11
- Date range: 2025-01-01 to 2025-06-29
- Regions: North, South, West
- Treated region: North
- Intervention date: 2025-05-01
- Comparison regions: South and West

## Data Preparation
The original dataset was transformed into two regional revenue tables.

### Daily Revenue Table
File: `daily_wide_revenue.csv`

- 180 rows and 4 columns
- Columns: date, North, South, West
- Each row represents one date.
- Regional values are summed recorded revenue.

### Weekly Revenue Table
File: `weekly_wide_revenue.csv`

- 26 rows and 4 columns
- Columns: week_start, North, South, West
- Each row represents one Monday-starting week.
- Regional values are summed recorded revenue.

## Validation
- Daily table: 180 rows, no missing values, no duplicate dates.
- Weekly table: 26 rows, no missing values, no duplicate weeks.
- Both tables are prepared for further analysis.

## Planned Analysis
1. Exploratory Data Analysis (EDA)
2. Synthetic Control Method (SCM)
3. Mean Squared Prediction Error (MSPE) validation
4. Double Machine Learning (DML)
5. Difference-in-Differences (DiD)
6. Comparison and interpretation of results

## Important Note
The dataset is synthetic and does not represent real company performance.
Any estimated treatment effects are based on the simulated data.
"""

with open("README.md", "w", encoding="utf-8") as file:
    file.write(readme_content.strip() + "\n")

print("README.md created successfully!")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("project1_algorithmic_pricing_synthetic_dataset.csv")
df["date"] = pd.to_datetime(df["date"])

intervention_date = pd.Timestamp("2025-05-01")

print("Dataset start:", df["date"].min())
print("Dataset end:", df["date"].max())
print("Intervention date:", intervention_date)

expected_treatment = (
    (df["region"] == "North") &
    (df["date"] >= intervention_date)
).astype(int)

print(
    "Treatment mismatches:",
    (df["treatment"] != expected_treatment).sum()
)
```

```python
df["period"] = np.where(
    df["date"] < intervention_date,
    "pre_treatment",
    "post_treatment"
)

pre_treatment = df[df["date"] < intervention_date].copy()
post_treatment = df[df["date"] >= intervention_date].copy()

print("Pre-treatment rows:", len(pre_treatment))
print("Post-treatment rows:", len(post_treatment))

display(
    df.groupby(["period", "region"])
      .agg(
          rows=("revenue", "size"),
          total_revenue=("revenue", "sum"),
          average_revenue=("revenue", "mean")
      )
      .reset_index()
)
```

```python
pre_treatment.to_csv("pre_treatment_data.csv", index=False)
post_treatment.to_csv("post_treatment_data.csv", index=False)

print("Treatment period datasets saved.")
```

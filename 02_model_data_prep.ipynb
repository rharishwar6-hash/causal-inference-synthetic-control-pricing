```python
import pandas as pd

df = pd.read_csv("project1_algorithmic_pricing_synthetic_dataset.csv")
df["date"] = pd.to_datetime(df["date"])

print("Dataset shape:", df.shape)
display(df.head())
```

```python
daily_wide = (
    df.pivot_table(
        index="date",
        columns="region",
        values="revenue",
        aggfunc="sum"
    )
    .reset_index()
)

daily_wide.columns.name = None

display(daily_wide.head(10))
print("Table shape:", daily_wide.shape)
```

```python
weekly_wide = (
    df.assign(
        week_start=df["date"]
        - pd.to_timedelta(df["date"].dt.dayofweek, unit="D")
    )
    .pivot_table(
        index="week_start",
        columns="region",
        values="revenue",
        aggfunc="sum"
    )
    .reset_index()
)

weekly_wide.columns.name = None

display(weekly_wide.head(10))

print("Weekly table shape:", weekly_wide.shape)
print("Unique weeks:", weekly_wide["week_start"].nunique())
```

```python
# Save daily wide table
daily_wide.to_csv("daily_wide_revenue.csv", index=False)

# Save weekly wide table
weekly_wide.to_csv("weekly_wide_revenue.csv", index=False)

print("Both CSV files saved successfully!")
```

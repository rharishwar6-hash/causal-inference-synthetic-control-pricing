```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("project1_algorithmic_pricing_synthetic_dataset.csv")

df["date"] = pd.to_datetime(df["date"])

df.head()
```

```python
region_revenue = df.groupby("region")["revenue"].sum().sort_values(
    ascending=False
)

display(region_revenue)

region_revenue.plot(kind="bar", figsize=(8, 5))

plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
price_comparison = df[
    ["our_price", "competitor_price"]
].mean()

display(price_comparison)

price_comparison.plot(kind="bar", figsize=(7, 5))

plt.title("Average Our Price vs Competitor Price")
plt.ylabel("Average Price")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
daily_revenue = df.groupby("date")["revenue"].sum()

plt.figure(figsize=(12, 5))
plt.plot(daily_revenue.index, daily_revenue.values)

plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Total Daily Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("EDA Summary")
print("Total rows:", df.shape[0])
print("Total regions:", df["region"].nunique())
print("Highest revenue region:", region_revenue.idxmax())
print("Average our price:", round(df["our_price"].mean(), 2))
print("Average competitor price:", round(df["competitor_price"].mean(), 2))
print("Daily revenue records:", daily_revenue.shape[0])
```
